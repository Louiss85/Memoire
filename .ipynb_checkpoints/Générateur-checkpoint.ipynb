##  <span style="color: #FFFFE0; font-weight: bold;">Création d'un lexique de formes fléchies pour chaque auteur</span>

```python
import os
import pandas as pd
import re
import shutil
import time

# Chemins des dossiers
correspondances_folders = [
    r"C:\\Users\\louis\\OneDrive - Université de Poitiers\\Fac\\Master\\Mémoire\\Variation flexions\\Textes locuteurs\\Poitevin\\Formes trouvées\\Correspondances",
    r"C:\\Users\\louis\\OneDrive - Université de Poitiers\\Fac\\Master\\Mémoire\\Variation flexions\\Textes locuteurs\\Saintongeais\\Formes trouvées\\Correspondances"
]
terminaisons_folder = r"C:\\Users\\louis\\OneDrive - Université de Poitiers\\Fac\\Master\\Mémoire\\Variation flexions\\Localiser locuteurs\\Terminaisons"
additional_terminaisons_folder = r"C:\\Users\\louis\\OneDrive - Université de Poitiers\\Fac\\Master\\Mémoire\\Variation flexions\\Localiser locuteurs\\Terminaisons déduites"
additional_folders = [
    r"C:\\Users\\louis\\OneDrive - Université de Poitiers\\Fac\\Master\\Mémoire\\Variation flexions\\Textes locuteurs\\Poitevin\\Lexique locuteur\\Lexiques nettoyés",
    r"C:\\Users\\louis\\OneDrive - Université de Poitiers\\Fac\\Master\\Mémoire\\Variation flexions\\Textes locuteurs\\Saintongeais\\Lexique locuteur\\Lexiques nettoyés",
    r"C:\\Users\\louis\\OneDrive - Université de Poitiers\\Fac\\Master\\Mémoire\\Variation flexions\\Emprunts\\Saintongeais\\Correspondances",
    r"C:\\Users\\louis\\OneDrive - Université de Poitiers\\Fac\\Master\\Mémoire\\Variation flexions\\Emprunts\\Poitevin\\Correspondances"
]
output_folder = r"C:\\Users\\louis\\OneDrive - Université de Poitiers\\Fac\\Master\\Mémoire\\Variation flexions\\Flexions\\Lexiques\\Formes régulières"
copy_folder = r"C:\\Users\\louis\\OneDrive - Université de Poitiers\\Fac\\Master\\Mémoire\\Variation flexions\\Flexions\\Lexiques\\Reg + Ireg"

# Liste des noms de fichiers cibles pour la génération des variantes avec apostrophe
target_files = [
    "Eugène Charier", "Yves Rabault", "Marc Jean-Baptiste Cerisier", "Eugène Rougier",
    "Ernest Guyonnet", "Guy Marquais", "Georges Belluteau", "Daniel Abran", "Gérard Sansey",
    "Suzanne Bontems", "Roger Prieur", "Jean-Henri Burgaud", "Jean Naud", "Lucien Fabre-Bobard",
    "Marius Gagnère", "Adolphe Métivier", "Périnaud", "Pierre Péronneau", "Marc-Henri Evariste",
    "Odette Chapouthier", "Germaine Morand", "Lucien Racinoux", "Jean Daviaud", "Gilbert Tanneau",
    "Arsène Garnier", "René-Marie Lacuve", "Roger Soulard", "Françoise Couton",
    "Renée Gautier", "Emile Gauvrit", "Jean et Marguerite Deguimps", "Germaine Elie", "Alphonse Favrault",
    "Alphonse Gauvrit", "André Pacher", "Claude Burneau", "Edouard Lacuve", "Eloi Guitteny", "Henri Martin",
    "Jacques Pingot", "Louis Desmier", "Pierre Chevrier", "Adémien Lagarde", "Alexandre Hus", "Alexandre Négret",
    "Dominique Patenbois", "Dorothy Michaud", "Marc Marchadier"
]

# Charger le fichier "dicopoitevin glose.tsv"
dicopoitevin_path = r"C:\\Users\\louis\\OneDrive - Université de Poitiers\\Fac\\Master\\Mémoire\\Variation flexions\\Dicopoitevin\\dicopoitevin glose.tsv"
dicopoitevin_df = pd.read_csv(dicopoitevin_path, delimiter='\t')

def collect_lemmes_from_all_sources(correspondances_path, additional_folders):
    correspondances_df = pd.read_csv(correspondances_path, delimiter='\t')
    if 'Auteur' not in correspondances_df.columns or 'Dicopoitevin' not in correspondances_df.columns:
        raise ValueError("Les colonnes 'Auteur' ou 'Dicopoitevin' sont absentes du fichier de correspondances.")
    infinitifs = correspondances_df['Auteur'].tolist()
    supralemmas = [str(supralemma) for supralemma in correspondances_df['Dicopoitevin'].tolist()]
    base_filename = os.path.splitext(os.path.basename(correspondances_path))[0].replace('_O', '')
    for additional_folder in additional_folders:
        additional_file_path = os.path.join(additional_folder, base_filename + '.tsv')
        if os.path.exists(additional_file_path):
            try:
                additional_df = pd.read_csv(additional_file_path, delimiter='\t', on_bad_lines='warn')
                if 'lemme' not in additional_df.columns:
                    continue
                additional_lemmes = additional_df['lemme'].tolist()
                infinitifs.extend(additional_lemmes)
                supralemmas.extend(additional_lemmes)
            except pd.errors.ParserError:
                continue
    return infinitifs, supralemmas

def process_terminaisons(terminaisons_path, additional_terminaisons_folder):
    with open(terminaisons_path, 'r', encoding='utf-8') as file:
        terminaisons = file.readlines()
    terminaisons_dict = {}
    for ligne in terminaisons:
        if ligne.startswith('* '):
            elements = ligne.strip().split(' : ')
            if len(elements) == 2:
                type_terminaison, terminaisons_list = elements
                if type_terminaison in terminaisons_dict:
                    terminaisons_dict[type_terminaison].extend(terminaisons_list.split(', '))
                else:
                    terminaisons_dict[type_terminaison] = terminaisons_list.split(', ')
    if '* 1inf' not in terminaisons_dict:
        raise KeyError("La clé '* 1inf' est absente du fichier de terminaisons.")
    terminaisons_inf = sorted(terminaisons_dict['* 1inf'], key=len, reverse=True)
    additional_terminaisons_path = os.path.join(additional_terminaisons_folder, os.path.basename(terminaisons_path))
    if os.path.exists(additional_terminaisons_path):
        with open(additional_terminaisons_path, 'r', encoding='utf-8') as additional_file:
            additional_terminaisons = additional_file.readlines()
            for ligne in additional_terminaisons:
                if ligne.startswith('* '):
                    elements = ligne.strip().split(' : ')
                    if len(elements) == 2:
                        type_terminaison, terminaisons_list = elements
                        if type_terminaison in terminaisons_dict:
                            terminaisons_dict[type_terminaison].extend(terminaisons_list.split(', '))
                        else:
                            terminaisons_dict[type_terminaison] = terminaisons_list.split(', ')
    return terminaisons_dict, terminaisons_inf

# Nouvelle règle de génération des variantes avec apostrophe
# → renvoie un dict {forme_variante: contexte}
def generate_variants_with_context(forme):
    variants = {forme: None}
    chars = list(forme)

    # 1) Règle spéciale : "com" -> "c'm"
    i = 0
    while i < len(chars) - 2:
        if chars[i] == 'c' and chars[i+1] == 'o' and chars[i+2] == 'm':
            new_chars = chars.copy()
            new_chars[i+1] = "'"
            new_form = ''.join(new_chars)
            variants[new_form] = "com"
        i += 1

    # 2) Règles sur "e" : seulement si le "e" est encadré par les contextes spécifiés
    i = 0
    while i < len(chars):
        if chars[i] == 'e':
            # Contexte gauche et droit d'un caractère
            if i-1 >= 0 and i+1 < len(chars):
                left = chars[i-1]
                right = chars[i+1]
                if (left, right) in [
                    ('v', 'n'),
                    ('m', 'n'),
                    ('t', 'n'),
                    ('v', 'l'),
                    ('g', 'l'),
                    ('l', 'v'),
                    ('d', 'v'),
                    ('r', 'g'),
                    ('r', 'p'),
                    ('d', 'm'),
                ]:
                    new_chars = chars.copy()
                    new_chars[i] = "'"
                    new_form = ''.join(new_chars)

                    # on encode le contexte avec la voyelle (comme com / ven)
                    trigram = f"{left}{chars[i]}{right}"
                    variants[new_form] = trigram

            # Cas particulier "qu'm" : contexte gauche "qu", contexte droit "m"
            if i-2 >= 0 and ''.join(chars[i-2:i]) == 'qu' and i+1 < len(chars) and chars[i+1] == 'm':
                new_chars = chars.copy()
                new_chars[i] = "'"
                new_form = ''.join(new_chars)
                variants[new_form] = "qum"
        i += 1

    # on ne renvoie que les variantes (sans la forme de base)
    return {f: ctx for f, ctx in variants.items() if f != forme}

def generate_forms(infinitifs, supralemmas, terminaisons_dict, terminaisons_inf, terminaisons_filename, dialect, apply_apostrophe_variants=False):
    base_verbes = {}
    for infinitif in infinitifs:
        for terminaison in terminaisons_inf:
            if str(infinitif).endswith(terminaison):
                base_verbes[infinitif] = infinitif[:-len(terminaison)]
                break
        else:
            base_verbes[infinitif] = infinitif

    resultats = []

    # tense = valeur temporelle, mode = catégorie (Indicatif, Subjonctif, etc.)
    tense_mapping = {
        '* 1pas': 'Passé simple',
        '* 1fut': 'Futur',
        '* 1imp': 'Imparfait',
        '* 1subi': 'Imparfait',
        '* 1pp': 'Passé',
        '* 1pst': 'Présent',
        '* 1ppst': 'Présent',
        '* 1impe': 'Présent',
        '* 1subp': 'Présent',        # ← ici : Subjonctif présent
        '* 1inf': 'Présent',
        '* 1cond': 'Présent',
        '* 1comp': 'Passé composé',
    }
    mode_mapping = {
        '* 1pas': 'Indicatif',
        '* 1fut': 'Indicatif',
        '* 1imp': 'Indicatif',
        '* 1subi': 'Subjonctif',
        '* 1pp': 'Participe',
        '* 1pst': 'Indicatif',
        '* 1ppst': 'Participe',
        '* 1impe': 'Impératif',
        '* 1subp': 'Subjonctif',
        '* 1inf': 'Infinitif',
        '* 1cond': 'Conditionnel',
        '* 1comp': 'Indicatif',
    }

    genders_pp = ['m', 'f', 'm', 'f']
    persons = [1, 2, 3, 1, 2, 3]

    def determine_person(type_terminaison, index):
        if type_terminaison in ['* 1pp', '* 1ppst', '* 1inf', '* 1comp']:
            return ''
        elif type_terminaison == '* 1impe':
            return [2, 1, 2][index % 3]
        else:
            return persons[index % len(persons)]

    def determine_number(type_terminaison, index):
        if type_terminaison == '* 1imp':
            return ['sg', 'sg', 'sg', 'pl', 'pl', 'pl'][index % 6]
        elif type_terminaison == '* 1pp':
            return ['sg', 'sg', 'pl', 'pl'][index % 4]
        elif type_terminaison in ['* 1ppst', '* 1inf', '* 1comp']:
            return ''
        elif type_terminaison == '* 1impe':
            return ['sg', 'pl', 'pl'][index % 3]
        else:
            return ['sg', 'sg', 'sg', 'pl', 'pl', 'pl'][index % 6]

    for idx, infinitif in enumerate(infinitifs):
        supralemma = supralemmas[idx] if idx < len(supralemmas) else ''
        base_verbale = base_verbes[infinitif]

        # Ligne pour l'infinitif lui-même
        resultats.append({
            'deducted form': '',
            'form': infinitif,
            'lemma': infinitif,
            'supralemma': supralemma,
            'pos': 'VERB',
            'dialect': dialect,
            'glose': '',
            'tense': tense_mapping.get('* 1inf', ''),
            'mode': mode_mapping.get('* 1inf', ''),
            'person': '',
            'number': '',
            'gender': '',
            'variant': '',
            'comment': '',
            'source': terminaisons_filename
        })

        for type_terminaison, terminaisons in terminaisons_dict.items():
            if type_terminaison != '* 1inf':
                for index, terminaison in enumerate(terminaisons):
                    if terminaison != 'X':
                        # Forme de base
                        forme = base_verbale + terminaison
                        if (base_verbale.endswith('g') or base_verbale.endswith('gh')) and terminaison.startswith(('a', 'o')):
                            forme = base_verbale + 'e' + terminaison

                        tense = tense_mapping.get(type_terminaison, '')
                        mode = mode_mapping.get(type_terminaison, '')
                        person = determine_person(type_terminaison, index)
                        number = determine_number(type_terminaison, index)
                        gender = genders_pp[index % len(genders_pp)] if type_terminaison == '* 1pp' else ''

                        source = terminaisons_filename if mode == 'Infinitif' else 'Louis Lièvre'
                        if 'x' not in forme:
                            source = re.sub(r'\d+', '', terminaisons_filename)

                        # --- ALTERNANCES GRAPHIQUES EXCLUSIVES ---
                        forme_a_ajouter = forme
                        commentaire = ''

                        # c → qu
                        if base_verbale.endswith('c') and not terminaison.startswith('a') \
                           and str(supralemma).endswith(('càe', 'cér')):
                            forme_a_ajouter = base_verbale[:-1] + 'qu' + terminaison
                            commentaire = 'Transformation de c en qu'

                        # g → gu
                        elif base_verbale.endswith('g') and not terminaison.startswith('a') \
                             and str(supralemma).endswith('gàe'):
                            forme_a_ajouter = base_verbale[:-1] + 'gu' + terminaison
                            commentaire = 'Transformation de g en gu'

                        # gh → ghu
                        elif base_verbale.endswith('gh') and not terminaison.startswith('a') \
                             and str(supralemma).endswith('gàe'):
                            forme_a_ajouter = base_verbale[:-2] + 'ghu' + terminaison
                            commentaire = 'Transformation de gh en ghu'

                        # ç → c
                        elif base_verbale.endswith('ç') and not terminaison.startswith('a') \
                             and str(supralemma).endswith('çàe'):
                            forme_a_ajouter = base_verbale[:-1] + 'c' + terminaison
                            commentaire = 'Transformation de ç en c'

                        # On ajoute une seule forme (base ou alternée)
                        if forme_a_ajouter and 'x' not in forme_a_ajouter:
                            # forme sans apostrophe
                            resultats.append({
                                'deducted form': '',
                                'form': forme_a_ajouter,
                                'lemma': infinitif,
                                'supralemma': supralemma,
                                'pos': 'VERB',
                                'dialect': dialect,
                                'glose': '',
                                'tense': tense,
                                'mode': mode,
                                'person': person,
                                'number': number,
                                'gender': gender,
                                'variant': '',
                                'comment': commentaire,
                                'source': source
                            })

                            # variantes avec apostrophe + commentaire
                            if apply_apostrophe_variants:
                                variants_with_ctx = generate_variants_with_context(forme_a_ajouter)
                                for variant, ctx in variants_with_ctx.items():
                                    if variant != forme_a_ajouter:
                                        if ctx is not None:
                                            ctx_str = f"<{ctx}>"
                                        else:
                                            ctx_str = ""
                                        comment_variant = "Forme alternative avec apostrophe selon le contexte"
                                        if ctx_str:
                                            comment_variant += f" {ctx_str}"

                                        resultats.append({
                                            'deducted form': '',
                                            'form': variant,
                                            'lemma': infinitif,
                                            'supralemma': supralemma,
                                            'pos': 'VERB',
                                            'dialect': dialect,
                                            'glose': '',
                                            'tense': tense,
                                            'mode': mode,
                                            'person': person,
                                            'number': number,
                                            'gender': gender,
                                            'variant': '',
                                            'comment': comment_variant,
                                            'source': source
                                        })

    return resultats

def supprimer_brackets(texte):
    if isinstance(texte, str):
        texte = re.sub(r'\[.*?\]', '', texte)
        texte = re.sub(r'\(.*?\)', '', texte)
        texte = re.sub(r'\b[123]\.\s*', '', texte)
        texte = texte.strip()
        return texte if texte else ''
    return ''

def filter_specific_verbs(resultats_df):
    specific_verbs = ["neiger", "bruire", "sourdre à nouveau", "bruiner", "frire", "sourdre", "crachiner", "pleuvoir légèrement", "pleuvoir abondamment"]
    filtered_indices = set()
    for index, row in resultats_df.iterrows():
        if row['supralemme2'] in specific_verbs:
            if not (row['person'] == 3 and row['number'] in ['sg']):
                filtered_indices.add(index)
    return resultats_df.drop(filtered_indices)

def copy_generated_files(output_folder, copy_folder, filename):
    if not os.path.exists(copy_folder):
        os.makedirs(copy_folder)
    src_file = os.path.join(output_folder, filename)
    dst_file = os.path.join(copy_folder, filename)
    try:
        shutil.copy2(src_file, dst_file)
    except PermissionError:
        print(f"⚠️ Impossible de copier {filename} (fichier verrouillé)")
        time.sleep(0.5)
        try:
            shutil.copy2(src_file, dst_file)
        except PermissionError:
            print(f"❌ Échec définitif pour {filename}")

total_files = len([f for f in os.listdir(terminaisons_folder) if f.endswith('.txt')])
processed_files = 0

for terminaisons_file in os.listdir(terminaisons_folder):
    if terminaisons_file.endswith('.txt'):
        terminaisons_path = os.path.join(terminaisons_folder, terminaisons_file)
        terminaisons_dict, terminaisons_inf = process_terminaisons(terminaisons_path, additional_terminaisons_folder)
        terminaisons_filename = os.path.splitext(os.path.basename(terminaisons_path))[0]
        apply_apostrophe_variants = terminaisons_filename in target_files
        correspondances_path = None
        for correspondances_folder in correspondances_folders:
            correspondances_candidate = os.path.join(correspondances_folder, terminaisons_filename + '_O.txt')
            if os.path.exists(correspondances_candidate):
                correspondances_path = correspondances_candidate
                break
        if correspondances_path:
            infinitifs, supralemmas = collect_lemmes_from_all_sources(correspondances_path, additional_folders)
            dialect = "Poitevin" if "Poitevin" in correspondances_path else "Saintongeais"
            resultats = generate_forms(infinitifs, supralemmas, terminaisons_dict, terminaisons_inf, terminaisons_filename, dialect, apply_apostrophe_variants)
            resultats_df = pd.DataFrame(resultats)
            for index, row in resultats_df.iterrows():
                if row['mode'] == 'Infinitif':
                    supralemma = row['supralemma']
                    glose_entry = dicopoitevin_df[dicopoitevin_df['forme'] == supralemma]
                    if glose_entry.empty:
                        for additional_folder in additional_folders:
                            lexique_path = os.path.join(additional_folder, terminaisons_filename + '.tsv')
                            if os.path.exists(lexique_path):
                                try:
                                    lexique_df = pd.read_csv(lexique_path, delimiter='\t')
                                    if 'lemme' in lexique_df.columns and 'glose' in lexique_df.columns:
                                        glose_entry = lexique_df[lexique_df['lemme'] == supralemma]
                                        if not glose_entry.empty:
                                            resultats_df.at[index, 'glose'] = glose_entry.iloc[0]['glose']
                                            break
                                except pd.errors.ParserError:
                                    continue
                    if not glose_entry.empty:
                        resultats_df.at[index, 'glose'] = glose_entry.iloc[0]['glose']
                        if 'source_entry' in glose_entry.columns:
                            source_entry = glose_entry.iloc[0]['source_entry']
                            if source_entry.endswith('.'):
                                source_entry = source_entry[:-1]
                            if source_entry.startswith("vt. = ") or source_entry.startswith("vi. = ") or source_entry.startswith("vp. = "):
                                forme = source_entry[5:].strip()
                            elif source_entry.startswith("vt/i. = ") or source_entry.startswith("vt/p. = "):
                                forme = source_entry[7:].strip()
                            else:
                                continue
                            glose_entry = dicopoitevin_df[dicopoitevin_df['forme'] == forme]
                            if not glose_entry.empty:
                                resultats_df.at[index, 'glose'] = glose_entry.iloc[0]['glose']
            resultats_df['glose'] = resultats_df['glose'].apply(supprimer_brackets)
            resultats_df['supralemme2'] = ""
            for index, row in resultats_df.iterrows():
                glose = row['glose']
                if isinstance(glose, str) and glose.strip():
                    supralemma = row['supralemma']
                    resultats_df.loc[resultats_df['supralemma'] == supralemma, 'supralemme2'] = glose
            resultats_df['glose'] = ""
            resultats_df = filter_specific_verbs(resultats_df)

            # === PATCH MINIMAL AMÉLIORÉ ===
            output_path = os.path.join(output_folder, f"{terminaisons_filename}.tsv")
            temp_path = os.path.join(output_folder, f"temp_{terminaisons_filename}.tsv")

            # 1. Écrire dans un fichier temporaire
            resultats_df.to_csv(temp_path, sep='\t', index=False, encoding='utf-8')

            # 2. Supprimer le fichier de destination s'il existe déjà
            if os.path.exists(output_path):
                try:
                    os.remove(output_path)
                except PermissionError:
                    print(f"⚠️ Impossible de supprimer {output_path} (fichier verrouillé)")
                    time.sleep(0.5)
                    try:
                        os.remove(output_path)
                    except PermissionError:
                        print(f"❌ Échec définitif pour la suppression de {output_path}")
                        continue

            # 3. Renommer le fichier temporaire
            try:
                os.rename(temp_path, output_path)
            except PermissionError:
                print(f"⚠️ Impossible de renommer {temp_path} en {output_path} (fichier verrouillé)")
                time.sleep(0.5)
                try:
                    os.rename(temp_path, output_path)
                except PermissionError:
                    print(f"❌ Échec définitif pour le renommage de {temp_path}")
                    continue

            # 4. Copier avec gestion d'erreur
            copy_generated_files(output_folder, copy_folder, f"{terminaisons_filename}.tsv")
            # === FIN PATCH ===

            processed_files += 1
            print(f"\r{processed_files}/{total_files} fichiers traités", end='', flush=True)

print("\nFini :)")
```

##  <span style="color: #FFFFE0; font-weight: bold;">Création d'un lexique des verbes irréguliers pour chaque auteur</span>

```python
import os
import pandas as pd
import re

# Chemin vers le fichier dicopoitevin
dicopoitevin_path = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Dicopoitevin\dicopoitevin glose.tsv"

# Charger le fichier TSV dans un DataFrame
dicopoitevin_df = pd.read_csv(dicopoitevin_path, sep='\t', encoding='utf-8')

# Vérifier que les colonnes nécessaires sont présentes
required_columns = ['forme', 'pos', 'glose']
if not all(column in dicopoitevin_df.columns for column in required_columns):
    raise ValueError(f"Le fichier doit contenir les colonnes suivantes : {required_columns}")

# Chemins des dossiers contenant les fichiers de terminaisons
terminaisons_folders = {
    "Etre_Poitevin": r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Etre\Terminaisons\Poitevin",
    "Etre_Saintongeais": r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Etre\Terminaisons\Saintongeais",
    "Dire_Poitevin": r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Dire\Terminaisons\Poitevin",
    "Dire_Saintongeais": r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Dire\Terminaisons\Saintongeais",
    "Avoir_Poitevin": r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Avoir\Terminaisons\Poitevin",
    "Avoir_Saintongeais": r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Avoir\Terminaisons\Saintongeais",
    "Voir_Poitevin": r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Voir\Terminaisons\Poitevin",
    "Voir_Saintongeais": r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Voir\Terminaisons\Saintongeais",
    "Faire_Saintongeais": r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Faire\Terminaisons\Saintongeais",
    "Faire_Poitevin": r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Faire\Terminaisons\Poitevin"
}

# Dossier de sortie pour les fichiers TSV
output_folder = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Lexiques générés"

# Dossier contenant les fichiers existants à fusionner
existing_folder = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Flexions\Lexiques\Reg + Ireg"

# Fonction pour fusionner les fichiers sans doublons
def merge_files(existing_path, new_path):
    # Vérifier si le fichier est vide
    if os.path.getsize(existing_path) == 0:
        return

    try:
        existing_df = pd.read_csv(existing_path, sep='\t', encoding='utf-8', low_memory=False)
    except pd.errors.EmptyDataError:
        return

    try:
        new_df = pd.read_csv(new_path, sep='\t', encoding='utf-8', low_memory=False)
    except pd.errors.EmptyDataError:
        return

    # Concaténer les deux DataFrames
    merged_df = pd.concat([existing_df, new_df], ignore_index=True)

    # Supprimer les doublons
    merged_df.drop_duplicates(inplace=True)

    # Écrire le fichier fusionné
    merged_df.to_csv(existing_path, sep='\t', index=False, encoding='utf-8')

# Fonction pour traiter un fichier de terminaisons
def process_terminaisons(terminaisons_path):
    with open(terminaisons_path, 'r', encoding='utf-8') as file:
        terminaisons = file.readlines()

    terminaisons_dict = {}
    infinitifs = []
    supralemme = ""

    for ligne in terminaisons:
        if ligne.startswith('* '):
            elements = ligne.strip().split(' : ')
            if len(elements) == 2:
                type_terminaison, terminaisons_list = elements
                if type_terminaison == '* 1inf' and terminaisons_list != 'X':
                    infinitifs.extend(terminaisons_list.split(', '))
                elif terminaisons_list != 'X':
                    if type_terminaison in terminaisons_dict:
                        terminaisons_dict[type_terminaison].extend(terminaisons_list.split(', '))
                    else:
                        terminaisons_dict[type_terminaison] = terminaisons_list.split(', ')
        elif ligne.startswith('supralemme = '):
            supralemme = ligne.split('=')[1].strip()

    return infinitifs, terminaisons_dict, supralemme

# Fonction pour créer un lexique à partir des terminaisons
def create_lexique(infinitifs, terminaisons_dict, supralemme, terminaisons_filename, dialect):
    resultats = []

    tense_mapping = {
        '* 1pas': 'Passé simple',
        '* 1fut': 'Futur',
        '* 1imp': 'Imparfait',
        '* 1subi': 'Imparfait',
        '* 1pp': 'Passé',
        '* 1pst': 'Présent',
        '* 1ppst': 'Présent',
        '* 1impe': 'Présent',
        '* 1subp': 'Présent',
        '* 1inf': 'Infinitif',
        '* 1cond': 'Présent',
        '* 1comp': 'Passé',
    }

    mode_mapping = {
        '* 1pas': 'Indicatif',
        '* 1fut': 'Indicatif',
        '* 1imp': 'Indicatif',
        '* 1subi': 'Subjonctif',
        '* 1pp': 'Participe',
        '* 1pst': 'Indicatif',
        '* 1ppst': 'Participe',
        '* 1impe': 'Impératif',
        '* 1subp': 'Subjonctif',
        '* 1inf': 'Infinitif',
        '* 1cond': 'Conditionnel',
        '* 1comp': 'Participe',
    }

    persons = [1, 2, 3, 1, 2, 3]

    def determine_person(type_terminaison, index):
        if type_terminaison in ['* 1pp', '* 1ppst', '* 1inf', '* 1comp']:
            return ''
        elif type_terminaison == '* 1impe':
            return [2, 1, 2][index % 3]
        else:
            return persons[index % len(persons)]

    def determine_number(type_terminaison, index):
        if type_terminaison == '* 1imp':
            return ['sg', 'sg', 'sg', 'pl', 'pl', 'pl'][index % 6]
        elif type_terminaison == '* 1pp':
            return ['sg', 'sg', 'pl', 'pl'][index % 4]
        elif type_terminaison in ['* 1ppst', '* 1inf', '* 1comp']:
            return ''
        elif type_terminaison == '* 1impe':
            return ['sg', 'pl', 'pl'][index % 3]
        else:
            return ['sg', 'sg', 'sg', 'pl', 'pl', 'pl'][index % 6]

    def determine_gender(type_terminaison, index):
        if type_terminaison == '* 1pp':
            return ['m', 'f', 'm', 'f'][index % 4]
        return ''

    # Ajouter les infinitifs s'ils existent
    for infinitif in infinitifs:
        resultats.append({
            'deducted form': '',
            'form': infinitif,
            'lemma': infinitif,
            'supralemma': supralemme,
            'pos': 'VERB',
            'dialect': dialect,
            'glose': '',
            'tense': 'Présent',
            'mode': mode_mapping.get('* 1inf', ''),
            'person': '',
            'number': '',
            'gender': '',
            'variant': '',
            'comment': '',
            'source': terminaisons_filename
        })

    # Ajouter les autres formes, même si l'infinitif est absent
    for type_terminaison, terminaisons in terminaisons_dict.items():
        for index, terminaison in enumerate(terminaisons):
            if terminaison != 'X':
                lemma = infinitifs[0] if infinitifs else ''
                gender = determine_gender(type_terminaison, index)
                resultats.append({
                    'deducted form': '',
                    'form': terminaison,
                    'lemma': lemma,
                    'supralemma': supralemme,
                    'pos': 'VERB',
                    'dialect': dialect,
                    'glose': '',
                    'tense': tense_mapping.get(type_terminaison, ''),
                    'mode': mode_mapping.get(type_terminaison, ''),
                    'person': determine_person(type_terminaison, index),
                    'number': determine_number(type_terminaison, index),
                    'gender': gender if gender else '',
                    'variant': '',
                    'comment': '',
                    'source': terminaisons_filename
                })

    return resultats

# Dictionnaire pour accumuler les résultats par auteur
results_by_author = {}

# Traiter tous les fichiers de terminaisons pour chaque dialecte
for folder_key, terminaisons_folder in terminaisons_folders.items():
    dialect = "Poitevin" if "Poitevin" in folder_key else "Saintongeais"
    for terminaisons_file in os.listdir(terminaisons_folder):
        if terminaisons_file.endswith('.txt'):
            terminaisons_path = os.path.join(terminaisons_folder, terminaisons_file)
            infinitifs, terminaisons_dict, supralemme = process_terminaisons(terminaisons_path)

            terminaisons_filename = os.path.splitext(os.path.basename(terminaisons_path))[0]

            resultats = create_lexique(infinitifs, terminaisons_dict, supralemme, terminaisons_filename, dialect)

            author = terminaisons_filename
            if author not in results_by_author:
                results_by_author[author] = []
            results_by_author[author].extend(resultats)

# Fonction pour supprimer le texte entre crochets et parenthèses dans une chaîne.
def supprimer_brackets(texte):
    # On vérifie si le texte est une chaîne de caractères.
    if isinstance(texte, str):
        # On utilise une expression régulière pour supprimer tout texte entre crochets [].
        texte = re.sub(r'\[.*?\]', '', texte)
        # On utilise une expression régulière pour supprimer tout texte entre parenthèses ().
        texte = re.sub(r'\(.*?\)', '', texte)
        # On utilise une expression régulière pour supprimer "1.", "2.", et "3."
        texte = re.sub(r'\b[123]\.\s*', '', texte)
        # On enlève les espaces superflus au début et à la fin de la chaîne.
        texte = texte.strip()
        # On retourne le texte nettoyé, ou une chaîne vide s'il ne reste rien.
        return texte if texte else ''
    # Si le texte n'est pas une chaîne, on retourne une chaîne vide.
    return ''

# Écrire les résultats dans des fichiers TSV par auteur et fusionner avec les fichiers existants
for author, results in results_by_author.items():
    results_df = pd.DataFrame(results)

    # Vérifiez si la colonne "glose" existe dans le DataFrame
    if 'glose' not in results_df.columns:
        results_df['glose'] = ""  # Ajoutez la colonne avec des valeurs vides si elle n'existe pas

    # Mettre à jour la colonne "glose"
    for index, row in results_df.iterrows():
        if row['lemma']:
            supralemma = row['lemma']
        else:
            supralemma = row['supralemma']

        glose_entry = dicopoitevin_df[(dicopoitevin_df['forme'] == supralemma) & (dicopoitevin_df['pos'] == 'VERB')]
        if not glose_entry.empty:
            # Ajoutez une condition pour vérifier le mode avant de mettre à jour la colonne "glose"
            if row['mode'] == 'Infinitif' and row['supralemma'] == 'dire':
                results_df.at[index, 'glose'] = glose_entry.iloc[0]['glose']

            # Vérifier systématiquement la colonne "source_entry" pour des formes alternatives
            if 'source_entry' in glose_entry.columns:
                source_entry = glose_entry.iloc[0]['source_entry']

                # Supprimer le point final si présent
                if source_entry.endswith('.'):
                    source_entry = source_entry[:-1]

                if source_entry.startswith("vt. = ") or source_entry.startswith("vi. = ") or source_entry.startswith("vp. = "):
                    forme = source_entry[5:].strip()
                elif source_entry.startswith("vt/i. = ") or source_entry.startswith("vt/p. = "):
                    forme = source_entry[7:].strip()
                else:
                    continue  # Passer à la ligne suivante si aucune condition n'est remplie

                # Chercher cette nouvelle forme dans la colonne "forme" de dicopoitevin_df
                glose_entry = dicopoitevin_df[(dicopoitevin_df['forme'] == forme) & (dicopoitevin_df['pos'] == 'VERB')]
                if not glose_entry.empty:
                    results_df.at[index, 'glose'] = glose_entry.iloc[0]['glose']

    # Appliquer la fonction "supprimer_brackets" à chaque élément de la colonne "glose"
    results_df['glose'] = results_df['glose'].apply(supprimer_brackets)

    # Ajouter la colonne "supralemme2" et la mettre à jour
    results_df['supralemme2'] = ""
    for index, row in results_df.iterrows():
        supralemme = row['supralemma']  # On récupère la valeur de la colonne "supralemme" pour la ligne actuelle.

        # Ajouter la condition pour "être"
        if supralemme == "étre":
            results_df.loc[results_df['supralemma'] == supralemme, 'supralemme2'] = "être"
        if supralemme == "dire":
            results_df.loc[results_df['supralemma'] == supralemme, 'supralemme2'] = "dire"
        if supralemme == "voer":
            results_df.loc[results_df['supralemma'] == supralemme, 'supralemme2'] = "voir"
        if supralemme == "avàe":
            results_df.loc[results_df['supralemma'] == supralemme, 'supralemme2'] = "avoir"
        if supralemme == "faere":
            results_df.loc[results_df['supralemma'] == supralemme, 'supralemme2'] = "faire"
        else:
            glose = row['glose']  # On récupère la valeur de la colonne "glose" pour la ligne actuelle.
            # On vérifie si "glose" est une chaîne non vide.
            if isinstance(glose, str) and glose.strip():
                # On met la valeur de "glose" dans la colonne "supralemme2" pour toutes les lignes où "supralemme" est égal à la valeur actuelle de "supralemme".
                results_df.loc[results_df['supralemma'] == supralemme, 'supralemme2'] = glose

    output_path = os.path.join(output_folder, f"{author}.tsv")
    results_df.to_csv(output_path, sep='\t', index=False, encoding='utf-8')

    existing_file_path = os.path.join(existing_folder, f"{author}.tsv")
    if os.path.exists(existing_file_path):
        merge_files(existing_file_path, output_path)
    else:
        results_df.to_csv(existing_file_path, sep='\t', index=False, encoding='utf-8')

print("Fini :)")
```

##  <span style="color: #FFFFE0; font-weight: bold;">Application d'une glose</span>

```python
import os
import pandas as pd
import re
from unidecode import unidecode

# Chemins des répertoires
input_directory = "C:/Users/louis/OneDrive - Université de Poitiers/Fac/Master/Mémoire/Variation flexions/Flexions/Lexiques/Reg + Ireg"
lexicon_path = "C:/Users/louis/OneDrive - Université de Poitiers/Fac/Master/Mémoire/Variation flexions/Lexiques/Lexique gloses2.tsv"
output_directory_with_gloses = "C:/Users/louis/OneDrive - Université de Poitiers/Fac/Master/Mémoire/Variation flexions/Flexions/Lexiques/Formes glosées/Avec"
output_directory_without_gloses = "C:/Users/louis/OneDrive - Université de Poitiers/Fac/Master/Mémoire/Variation flexions/Flexions/Lexiques/Formes glosées/Sans"

# Fonction pour nettoyer les mots des ponctuations et des préfixes pronominaux
def clean_word(word):
    # Supprimer le préfixe "s'"
    if isinstance(word, str) and word.startswith("s'"):
        word = word[2:]
    # Supprimer le préfixe "se "
    word = re.sub(r"^se ", "", str(word))
    # Supprimer les ponctuations
    word = re.sub(r"[^\w\s]", "", word)
    return word

# Fonction pour normaliser les mots en retirant les diacritiques
def normalize_word(word):
    return unidecode(str(word))

# Fonction pour appliquer un lexique
def apply_lexicon(groupe_df, lexicon_path):
    lexicon_df = pd.read_csv(lexicon_path, delimiter="\t")
    lexicon_dict = {}
    for _, lex_row in lexicon_df.iterrows():
        lemma = lex_row["lemme"]
        key = (
            lex_row["tense"],
            lex_row["mode"],
            lex_row["number"],
            lex_row["person"],
            lex_row["gender"],
        )
        lexicon_dict.setdefault(normalize_word(lemma), {})[key] = lex_row["glose"]

    new_rows = []
    modified_indices = set()

    for idx, groupe_row in groupe_df.iterrows():
        if pd.notna(groupe_row["glose"]):
            continue
        if pd.isna(groupe_row["supralemme2"]):
            continue

        # Condition pour "valter" -> "velter"
        if pd.notna(groupe_row["form"]) and str(groupe_row["form"]).strip().lower() == "valter":
            new_row = groupe_row.copy()
            new_row["glose"] = "velter"
            new_rows.append(new_row)
            modified_indices.add(idx)
            continue

        supralemmes = str(groupe_row["supralemme2"]).split()
        gloses = []

        for supralemme in supralemmes:
            supralemme = supralemme.strip()
            supralemme_clean = clean_word(supralemme)
            supralemme_normalized = normalize_word(supralemme_clean)

            if supralemme_normalized == "etorsae" and groupe_row["mode"] == "Infinitif":
                gloses.append("s'étirer")
                continue
            if supralemme_normalized == "enlisae" and groupe_row["mode"] == "Infinitif":
                gloses.append("s'enliser")
                continue

            if supralemme_normalized.endswith(("er", "re", "r")):
                if supralemme_normalized in lexicon_dict:
                    for key, g in lexicon_dict[supralemme_normalized].items():
                        if (
                            (pd.isna(groupe_row["tense"]) or pd.isna(key[0]) or groupe_row["tense"] == key[0])
                            and (pd.isna(groupe_row["mode"]) or pd.isna(key[1]) or groupe_row["mode"] == key[1])
                            and (pd.isna(groupe_row["number"]) or pd.isna(key[2]) or groupe_row["number"] == key[2])
                            and (pd.isna(groupe_row["person"]) or pd.isna(key[3]) or groupe_row["person"] == key[3])
                            and (pd.isna(groupe_row["gender"]) or pd.isna(key[4]) or groupe_row["gender"] == key[4])
                        ):
                            glose = g
                            gloses.append(glose)

        if gloses:
            modified_indices.add(idx)
            for glose in gloses:
                new_row = groupe_row.copy()
                new_row["glose"] = glose
                new_rows.append(new_row)

    if new_rows:
        new_rows_df = pd.DataFrame(new_rows)
        groupe_df = pd.concat([groupe_df, new_rows_df], ignore_index=True)

    return groupe_df, modified_indices

# Variables pour stocker le nombre total de lignes
total_with_gloses = 0
total_without_gloses = 0

# S'assurer que les dossiers de sortie existent
os.makedirs(output_directory_with_gloses, exist_ok=True)
os.makedirs(output_directory_without_gloses, exist_ok=True)

# Parcourir tous les fichiers du répertoire d'entrée
for filename in os.listdir(input_directory):
    if filename.endswith(".tsv"):
        main_file_path = os.path.join(input_directory, filename)
        groupe_df = pd.read_csv(main_file_path, delimiter="\t", low_memory=False)

        # Appliquer le lexique
        groupe_df, modified_indices = apply_lexicon(groupe_df, lexicon_path)

        # Séparer le fichier en deux
        with_gloses_df = groupe_df[groupe_df["glose"].notna()].copy()
        without_gloses_df = groupe_df[groupe_df["glose"].isna() & ~groupe_df.index.isin(modified_indices)].copy()

        # Supprimer les doublons
        with_gloses_df.drop_duplicates(inplace=True)
        without_gloses_df.drop_duplicates(inplace=True)

        # Supprimer la colonne "supralemme2" uniquement dans le fichier avec les gloses
        if "supralemme2" in with_gloses_df.columns:
            with_gloses_df.drop(columns=["supralemme2"], inplace=True)

        # Chemins de sortie
        base_filename = os.path.splitext(filename)[0]
        output_file_path_with_gloses = os.path.join(output_directory_with_gloses, f"{base_filename}.tsv")
        output_file_path_without_gloses = os.path.join(output_directory_without_gloses, f"{base_filename}.tsv")

        # Sauvegarder les fichiers modifiés
        with_gloses_df.to_csv(output_file_path_with_gloses, sep="\t", index=False)

        # Vérifier si le DataFrame sans gloses n'est pas vide avant de le sauvegarder
        if not without_gloses_df.empty:
            without_gloses_df.to_csv(output_file_path_without_gloses, sep="\t", index=False)

        # Calculer le nombre de non-correspondances
        non_correspondance_count = len(without_gloses_df)

        # Imprimer les résultats
        print(f"{filename} - {len(with_gloses_df)} / {non_correspondance_count}")

        # Mettre à jour les totaux (si tu voulais vraiment exclure la première ligne, garde le -1)
        total_with_gloses += max(len(with_gloses_df) - 1, 0)
        total_without_gloses += max(len(without_gloses_df) - 1, 0)

# Fusionner tous les fichiers avec gloses en un seul fichier (optionnel)
all_with_gloses_files = [
    os.path.join(output_directory_with_gloses, f)
    for f in os.listdir(output_directory_with_gloses)
    if f.endswith(".tsv")
]
if all_with_gloses_files:
    all_with_gloses_df = pd.concat(
        [pd.read_csv(f, delimiter="\t") for f in all_with_gloses_files],
        ignore_index=True,
    )
    all_with_gloses_df.drop_duplicates(inplace=True)
    # Si tu veux un fichier global, décommente :
    # all_with_gloses_df.to_csv(
    #     os.path.join(output_directory_with_gloses, "Lexique_avec.tsv"),
    #     sep="\t",
    #     index=False,
    # )

# Fusionner tous les fichiers sans gloses en un seul fichier
all_without_gloses_files = [
    os.path.join(output_directory_without_gloses, f)
    for f in os.listdir(output_directory_without_gloses)
    if f.endswith(".tsv")
]
if all_without_gloses_files:
    all_without_gloses_df = pd.concat(
        [pd.read_csv(f, delimiter="\t") for f in all_without_gloses_files],
        ignore_index=True,
    )
    all_without_gloses_df.drop_duplicates(inplace=True)
    all_without_gloses_df.to_csv(
        os.path.join(output_directory_without_gloses, "Lexique_sans.tsv"),
        sep="\t",
        index=False,
    )

# Imprimer le nombre total de lignes
print(f"Total avec gloses : {total_with_gloses}")
print(f"Total sans gloses : {total_without_gloses}")

# Calculer le pourcentage de lignes avec gloses
total_lines = total_with_gloses + total_without_gloses
percentage_with_gloses = (total_with_gloses / total_lines) * 100 if total_lines > 0 else 0
print(f"Pourcentage avec gloses : {percentage_with_gloses:.2f}%")
print("Fini :)")
```

##  <span style="color: #FFFFE0; font-weight: bold;">Forme déduite ou non</span>

```python
import os
import pandas as pd
import re

# Chemins des dossiers
lexiques_folder = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Flexions\Lexiques\Formes glosées\Avec"
textes_folders = [
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Textes locuteurs\Saintongeais\Textes en txt",
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Textes locuteurs\Poitevin\Textes en txt"
]
output_folder = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Flexions\Lexiques\Formes déduites"

os.makedirs(output_folder, exist_ok=True)

# Fonction pour lire un fichier texte en ANSI
def read_ansi_file(file_path):
    with open(file_path, 'r', encoding='ansi') as f:
        return f.read().lower()

# Précharger tous les fichiers texte pour éviter de les relire plusieurs fois
texte_cache = {}
for textes_folder in textes_folders:
    for file_name in os.listdir(textes_folder):
        if file_name.endswith(".txt"):
            full_path = os.path.join(textes_folder, file_name)
            texte_cache[file_name] = read_ansi_file(full_path)

# Liste des fichiers lexiques
lexique_files = [f for f in os.listdir(lexiques_folder) if f.endswith(".tsv")]
total_files = len(lexique_files)

# Liste des articles à ignorer
articles = {'le', 'les', 'lés', 'la', 'des', 'dés', 'un', 'une', 'in', 'ine', 'ma', 'ta', 'ton', 'tés', 'tes', 'sa', 'son', 'dau', 'du'}

for processed_files, lexique_file in enumerate(lexique_files, start=1):
    lexique_path = os.path.join(lexiques_folder, lexique_file)
    lexique_df = pd.read_csv(lexique_path, delimiter='\t', low_memory=False)

    base_name = os.path.splitext(lexique_file)[0]

    # Concaténer les textes correspondants depuis le cache
    texte_contents = " ".join(
        texte_cache[file_name]
        for file_name in texte_cache
        if file_name.startswith(base_name)
    )

    # Extraire tous les mots (ignorer ceux précédés d'apostrophe)
    words_list = re.findall(r"(?<!')\b\w+\b", texte_contents)
    
    words_in_text = set()
    for i, word in enumerate(words_list):
        # ignorer le mot si le précédent est un article
        if i > 0 and words_list[i-1].lower() in articles:
            continue
        words_in_text.add(word.lower())

    # Vérifier la présence des mots et remplir 'deducted form',
    # en ignorant spécifiquement certaines formes
    if 'form' in lexique_df.columns:
            lexique_df['deducted form'] = lexique_df['form'].apply(
                lambda w: '' if w.lower() in words_in_text and w.lower() not in {'affaires', 'conte', 'chevalaie', 'chasse', 'quoué', 'tchataite', 'rède',  'mouche', 'colère', 'fesses', 'farine', 'livre', 'branche', 'sale', 'guerre', 'linge', 'bonne', 'faute', 'feuilles', 'paroles', 'manche', 'méchant', 'parole', 'viande', 'vilajhe', 'gorét', 'pllume', 'voute', 'poules', 'gronde', 'gauche', 'feuille', 'poche', 'chance', 'huit", "pattes', 'çhéle', 'mine', 'noce', 'rené', 'affaire', 'machine', 'paille', 'cause', 'doute', 'force', 'garde', 'porte', 'soupe', 'marché', 'trois', 'route', 'portes', 'fali', 'petit', "mariée", "marié", "mariaie", "marie", "couté", "coûté"} else 'x'
        )

    # Mettre à jour la colonne 'source'
    if 'source' in lexique_df.columns:
        lexique_df.loc[lexique_df['deducted form'] == 'x', 'source'] = 'Louis Lièvre'
    else:
        lexique_df['source'] = lexique_df['deducted form'].apply(
            lambda form: 'Louis Lièvre' if form == 'x' else ''
        )

    # Sauvegarder le fichier
    output_file_name = lexique_file.replace("_avec", "")
    lexique_df.to_csv(os.path.join(output_folder, output_file_name), sep='\t', index=False)

    print(f"\r{processed_files}/{total_files} fichiers traités", end='', flush=True)

print("\nFini :)")
```

##  <span style="color: #FFFFE0; font-weight: bold;">Création du lexique général</span>

```python
import os
import pandas as pd

# Chemin du dossier contenant les fichiers TSV
input_folder = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Flexions\Lexiques\Formes déduites"

# Chemin de sortie pour le fichier combiné
output_path = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Flexions\FFPS2.0.tsv"

# Liste pour stocker les DataFrames de chaque fichier TSV
dataframes = []

# Parcourir tous les fichiers du dossier
for filename in os.listdir(input_folder):
    if filename.endswith('.tsv'):
        file_path = os.path.join(input_folder, filename)
        # Lire chaque fichier TSV dans un DataFrame avec low_memory=False
        df = pd.read_csv(file_path, delimiter='\t', low_memory=False)
        # Ajouter le DataFrame à la liste
        dataframes.append(df)

# Combiner tous les DataFrames en un seul
combined_df = pd.concat(dataframes, ignore_index=True)

# Écrire le DataFrame combiné dans un fichier TSV
combined_df.to_csv(output_path, sep='\t', index=False, encoding='utf-8')

print("Fini :)")
```
