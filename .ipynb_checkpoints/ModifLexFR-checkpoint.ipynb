##  <span style="color: #FFFFE0; font-weight: bold;">Fusion des trois lexiques pour n'en faire qu'un</span>

```python
import pandas as pd

# Chemins vers les fichiers TSV
file_paths = [
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Lexiques\Lexiques adaptés\analog.tsv",
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Lexiques\Lexiques adaptés\GLAFF.tsv",
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Lexiques\Lexiques adaptés\LEFFF.tsv"
]

# Lire chaque fichier TSV et les stocker dans une liste de DataFrames
dataframes = [pd.read_csv(file_path, delimiter='\t') for file_path in file_paths]

# Fusionner tous les DataFrames en un seul
merged_df = pd.concat(dataframes, ignore_index=True)

# Supprimer les doublons
merged_df.drop_duplicates(inplace=True)

# Chemin de sortie pour le fichier fusionné
output_file_path = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Lexiques\Lexique gloses1.tsv"

# Sauvegarder le DataFrame fusionné dans un nouveau fichier TSV
merged_df.to_csv(output_file_path, sep='\t', index=False)

print("Fichier créé")
```

##  <span style="color: #FFFFE0; font-weight: bold;">Correction des participes passés masculins singuliers sans accent</span>

```python
import pandas as pd

def update_glose(row):
    if row['gender'] == 'm' and row['number'] == 'sg' and row['glose'].endswith('e'):
        row['glose'] = row['glose'][:-1] + 'é'
    return row

def generate_missing_forms(row):
    lemme = row['lemme']
    glose = row['glose']
    pos = row['pos']
    tense = row['tense']
    mode = row['mode']
    new_rows = []
    if row['gender'] == 'm' and row['number'] == 'sg':
        base_participe = lemme.replace('er', 'é')
        feminin_sg = base_participe + 'e'
        masculin_pl = base_participe + 's'
        feminin_pl = base_participe + 'es'
        new_rows = [
            {'glose': feminin_sg, 'lemme': lemme, 'pos': pos, 'tense': tense, 'mode': mode, 'person': '', 'number': 'sg', 'gender': 'f'},
            {'glose': masculin_pl, 'lemme': lemme, 'pos': pos, 'tense': tense, 'mode': mode, 'person': '', 'number': 'pl', 'gender': 'm'},
            {'glose': feminin_pl, 'lemme': lemme, 'pos': pos, 'tense': tense, 'mode': mode, 'person': '', 'number': 'pl', 'gender': 'f'}
        ]
    return new_rows

# Charger le fichier TSV
file_path = "C:\\Users\\louis\\OneDrive - Université de Poitiers\\Fac\\Master\\Mémoire\\Variation flexions\\Lexiques\\Lexique gloses1.tsv"
df = pd.read_csv(file_path, sep='\t')

# Mettre à jour la colonne 'glose'
df = df.apply(update_glose, axis=1)

# Générer les nouvelles formes
new_rows = []
for index, row in df.iterrows():
    new_rows.extend(generate_missing_forms(row))

# Ajouter les nouvelles lignes
updated_df = pd.concat([df, pd.DataFrame(new_rows)], ignore_index=True)

# Supprimer les lignes indésirables :
# 1. glose finit par "ees" ou "ee" (tous genres)
# 2. glose finit par "es" ET gender = "m"
mask = (
    ~updated_df['glose'].str.endswith(('ees', 'ee'), na=False) &
    ~((updated_df['glose'].str.endswith('es', na=False)) & (updated_df['gender'] == 'm'))
)
updated_df = updated_df[mask]

# Sauvegarder le résultat
output_file_path = "C:\\Users\\louis\\OneDrive - Université de Poitiers\\Fac\\Master\\Mémoire\\Variation flexions\\Lexiques\\Lexique gloses2.tsv"
updated_df.to_csv(output_file_path, sep='\t', index=False)
print("Fini")
```
