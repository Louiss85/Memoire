## <span style="color: #FFFFE0; font-weight: bold;">Calcul des fréquences d'apparition des formes</span>

```python
import os
import pandas as pd
import re
from collections import Counter

# Chemins
tsv_directory = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Flexions\Lexiques\Formes déduites"
text_directories = [
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Textes locuteurs\Poitevin\Textes en txt",
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Textes locuteurs\Saintongeais\Textes en txt",
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Faire\Terminaisons\Saintongeais",
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Faire\Terminaisons\Poitevin",
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Avoir\Terminaisons\Saintongeais",
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Avoir\Terminaisons\Poitevin",
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Etre\Terminaisons\Saintongeais",
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Etre\Terminaisons\Poitevin",
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Voir\Terminaisons\Saintongeais",
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Voir\Terminaisons\Poitevin",
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Dire\Terminaisons\Saintongeais",
    r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Dire\Terminaisons\Poitevin"
]

# --------------------------------------------------
# Fonctions
# --------------------------------------------------

def count_words_in_file(file_path):
    with open(file_path, 'r', encoding='ansi') as f:
        text = f.read().lower()

    # Utiliser une regex pour capturer les mots non précédés de 'un', 'une', etc.
    words = re.findall(r'(?:^|\s)(?!un\s|une\s|in\s|la\s|le\s|de\s|des\s|mon\s|ma\s)(\w+)', text)
    return Counter(words)

def remove_digits(text):
    """Supprime les chiffres d'une chaîne."""
    return re.sub(r'\d+', '', text)

# --------------------------------------------------
# Pré-calcul des occurrences de tous les textes
# --------------------------------------------------

all_text_counts = {}  # {base_name: Counter()}

for text_directory in text_directories:
    for text_file in os.listdir(text_directory):
        if text_file.endswith('.txt'):
            base_name = os.path.splitext(text_file)[0]
            text_path = os.path.join(text_directory, text_file)

            if base_name not in all_text_counts:
                all_text_counts[base_name] = Counter()

            all_text_counts[base_name] += count_words_in_file(text_path)

# --------------------------------------------------
# Traitement des fichiers TSV
# --------------------------------------------------

for tsv_file in os.listdir(tsv_directory):
    if tsv_file.endswith('.tsv'):
        tsv_path = os.path.join(tsv_directory, tsv_file)
        df = pd.read_csv(tsv_path, delimiter='\t', low_memory=False)

        base_name = os.path.splitext(tsv_file)[0]
        source_name = remove_digits(base_name)

        # Calcul de la fréquence
        def get_frequency(row):
            if 'x' in str(row['deducted form']):
                return None
            word = str(row['form']).lower()
            return all_text_counts.get(base_name, Counter()).get(word, 0)

        df['frequence'] = df.apply(get_frequency, axis=1)
        df['frequence'] = df['frequence'].astype('object')

        # Marquer les formes non trouvées
        df.loc[df['frequence'] == 0.0, 'deducted form'] = 'x'
        df.loc[df['frequence'] == 0.0, 'frequence'] = ""

        # --------------------------------------------------
        # Source
        # --------------------------------------------------

        df['source'] = ""

        # Formes attestées → nom du texte (sans chiffres)
        df.loc[
            df['frequence'].apply(lambda x: isinstance(x, (int, float)) and x >= 1),
            'source'
        ] = source_name

        # Formes non attestées → Louis Lièvre
        df.loc[df['deducted form'] == 'x', 'source'] = "Louis Lièvre"

        # Sauvegarde
        df.to_csv(tsv_path, sep='\t', index=False)

print("Fini :)")
```

## <span style="color: #FFFFE0; font-weight: bold;">Calcul des fréquences en temps, mode, personne et nombre</span>

```python
import pandas as pd

# ------------------------------------------------------------------
# Chemins des fichiers
# ------------------------------------------------------------------
input_file = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Flexions\FFPS2.0.tsv"
output_dir = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Statistiques"

# ------------------------------------------------------------------
# Lecture du fichier TSV
# ------------------------------------------------------------------
data = pd.read_csv(input_file, delimiter='\t', low_memory=False)

# ------------------------------------------------------------------
# Filtrage des lignes sans "x" dans "deducted form"
# ------------------------------------------------------------------
filtered_data = data[data['deducted form'] != 'x']
total_forms = len(filtered_data)

# ------------------------------------------------------------------
# Fonction de formatage des pourcentages
# ------------------------------------------------------------------
def format_percentage(series):
    return series.apply(lambda x: f"{x:.2f}".replace('.', ',') + "%")

# ------------------------------------------------------------------
# Filtre commun : SANS Subjonctif / Infinitif / Impératif au Présent
# ------------------------------------------------------------------
filtered_data_sans_subj_inf_present = filtered_data[
    ~(
        (filtered_data['mode'].isin(['Subjonctif', 'Infinitif', 'Impératif'])) &
        (filtered_data['tense'] == 'Présent')
    )
]

total_forms_sans_subj_inf_present = len(filtered_data_sans_subj_inf_present)

# ------------------------------------------------------------------
# Filtre spécifique : SANS Participe (pour tous les tableaux)
# ------------------------------------------------------------------
filtered_data_sans_participe = filtered_data[filtered_data['mode'] != 'Participe']
filtered_data_sans_subj_inf_present_sans_participe = filtered_data_sans_subj_inf_present[
    filtered_data_sans_subj_inf_present['mode'] != 'Participe'
]

total_forms_sans_participe = len(filtered_data_sans_participe)
total_forms_sans_subj_inf_present_sans_participe = len(filtered_data_sans_subj_inf_present_sans_participe)

# ------------------------------------------------------------------
# 1. Tableau tense / mode (original)
# ------------------------------------------------------------------
tense_mode_stats = (
    filtered_data
    .groupby(['tense', 'mode'])
    .size()
    .reset_index(name='count')
)
tense_mode_stats['percentage'] = (tense_mode_stats['count'] / total_forms) * 100
tense_mode_stats['percentage'] = format_percentage(tense_mode_stats['percentage'])
tense_mode_stats_sorted = tense_mode_stats.sort_values(by='count', ascending=False)

with open(f"{output_dir}/tense_mode.txt", "w", encoding='utf-8') as f:
    f.write(tense_mode_stats_sorted.to_string(index=False, justify='left'))
    f.write(f"\n\nTotal des formes : {total_forms}")

# ------------------------------------------------------------------
# 2. Tableau tense / mode SANS Subjonctif / Infinitif / Impératif au Présent
# ------------------------------------------------------------------
tense_mode_stats_sans_subj_inf_present = (
    filtered_data_sans_subj_inf_present
    .groupby(['tense', 'mode'])
    .size()
    .reset_index(name='count')
)
tense_mode_stats_sans_subj_inf_present['percentage'] = (
    tense_mode_stats_sans_subj_inf_present['count'] / total_forms_sans_subj_inf_present
) * 100
tense_mode_stats_sans_subj_inf_present['percentage'] = format_percentage(
    tense_mode_stats_sans_subj_inf_present['percentage']
)
tense_mode_stats_sans_subj_inf_present_sorted = tense_mode_stats_sans_subj_inf_present.sort_values(by='count', ascending=False)

with open(f"{output_dir}/tense_mode_sans_subj_inf_imp.txt", "w", encoding='utf-8') as f:
    f.write(tense_mode_stats_sans_subj_inf_present_sorted.to_string(index=False, justify='left'))
    f.write(f"\n\nTotal des formes : {total_forms_sans_subj_inf_present}")

# ------------------------------------------------------------------
# 3. Tableau person / number SANS Participe
# ------------------------------------------------------------------
person_number_stats = (
    filtered_data_sans_participe
    .groupby(['person', 'number'])
    .size()
    .reset_index(name='count')
)
person_number_stats['percentage'] = (person_number_stats['count'] / total_forms_sans_participe) * 100
person_number_stats['percentage'] = format_percentage(person_number_stats['percentage'])
person_number_stats_sorted = person_number_stats.sort_values(by='count', ascending=False)

with open(f"{output_dir}/person_number.txt", "w", encoding='utf-8') as f:
    f.write(person_number_stats_sorted.to_string(index=False, justify='left'))
    f.write(f"\n\nTotal des formes : {total_forms_sans_participe}")

# ------------------------------------------------------------------
# 3bis. Tableau person / number SANS Subjonctif / Infinitif / Impératif au Présent ET SANS Participe
# ------------------------------------------------------------------
person_number_stats_sans_subj_inf_present_sans_participe = (
    filtered_data_sans_subj_inf_present_sans_participe
    .groupby(['person', 'number'])
    .size()
    .reset_index(name='count')
)
person_number_stats_sans_subj_inf_present_sans_participe['percentage'] = (
    person_number_stats_sans_subj_inf_present_sans_participe['count'] / total_forms_sans_subj_inf_present_sans_participe
) * 100
person_number_stats_sans_subj_inf_present_sans_participe['percentage'] = format_percentage(
    person_number_stats_sans_subj_inf_present_sans_participe['percentage']
)
person_number_stats_sans_subj_inf_present_sans_participe_sorted = (
    person_number_stats_sans_subj_inf_present_sans_participe
    .sort_values(by='count', ascending=False)
)

with open(
    f"{output_dir}/person_number_sans_subj_inf_imp.txt",
    "w",
    encoding="utf-8"
) as f:
    f.write(
        person_number_stats_sans_subj_inf_present_sans_participe_sorted
        .to_string(index=False, justify='left')
    )
    f.write(f"\n\nTotal des formes : {total_forms_sans_subj_inf_present_sans_participe}")

# ------------------------------------------------------------------
# 4. Tableau combiné tense / mode / person / number SANS Participe
# ------------------------------------------------------------------
combined_stats = (
    filtered_data_sans_participe
    .groupby(['tense', 'mode', 'person', 'number'])
    .size()
    .reset_index(name='count')
)
combined_stats['percentage'] = (combined_stats['count'] / total_forms_sans_participe) * 100
combined_stats['percentage'] = format_percentage(combined_stats['percentage'])
combined_stats_sorted = combined_stats.sort_values(by='count', ascending=False)

with open(f"{output_dir}/correlations.txt", "w", encoding='utf-8') as f:
    f.write(combined_stats_sorted.to_string(index=False, justify='left'))
    f.write(f"\n\nTotal des formes : {total_forms_sans_participe}")

# ------------------------------------------------------------------
# 5. Tableau combiné tense / mode / person / number SANS Subjonctif / Infinitif / Impératif au Présent ET SANS Participe
# ------------------------------------------------------------------
combined_stats_sans_subj_inf_present = (
    filtered_data_sans_subj_inf_present_sans_participe
    .groupby(['tense', 'mode', 'person', 'number'])
    .size()
    .reset_index(name='count')
)
combined_stats_sans_subj_inf_present['percentage'] = (
    combined_stats_sans_subj_inf_present['count'] / total_forms_sans_subj_inf_present_sans_participe
) * 100
combined_stats_sans_subj_inf_present['percentage'] = format_percentage(
    combined_stats_sans_subj_inf_present['percentage']
)
combined_stats_sans_subj_inf_present_sorted = combined_stats_sans_subj_inf_present.sort_values(by='count', ascending=False)

with open(
    f"{output_dir}/correlations_sans_subj_inf_imp.txt",
    "w",
    encoding='utf-8'
) as f:
    f.write(
        combined_stats_sans_subj_inf_present_sorted
        .to_string(index=False, justify='left')
    )
    f.write(
        f"\n\nTotal des formes : {total_forms_sans_subj_inf_present_sans_participe}"
    )

# ------------------------------------------------------------------
# Fin
# ------------------------------------------------------------------
print("Fini :)")
```

## <span style="color: #FFFFE0; font-weight: bold;">Indice de variabilité intralocuteur</span>

```python
import os
import pandas as pd

def calculer_indice_variabilite_et_variantes(fichier):
    formes_par_identifiant = {}
    total_formes = []
    variantes = set()

    with open(fichier, 'r', encoding='utf-8') as f:
        for ligne in f:
            if ligne.startswith('*'):
                parties = ligne.split(':')
                identifiant = parties[0].strip()
                partie_formes = parties[1].strip()
                liste_formes = [forme.strip() for forme in partie_formes.split(',') if forme.strip() != 'X']

                if identifiant not in formes_par_identifiant:
                    formes_par_identifiant[identifiant] = [set(liste_formes)]
                else:
                    formes_par_identifiant[identifiant].append(set(liste_formes))

                total_formes.extend(liste_formes)

    for identifiant, ensembles_formes in formes_par_identifiant.items():
        if len(ensembles_formes) > 1:
            reference = ensembles_formes[0]
            for formes in ensembles_formes[1:]:
                variantes.update(formes - reference)

    if len(total_formes) == 0:
        return 0, 0, 0  # Retourne 0 pour nb_variantes si total_formes est vide

    indice_variabilite = (len(variantes) / len(total_formes)) * 100
    return indice_variabilite, len(total_formes), len(variantes)  # Utilise len(variantes) au lieu de variantes

# Chemins vers les dossiers contenant les fichiers
dossier_localiser = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Localiser locuteurs\Terminaisons"
dossier_poitevin_faire = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Faire\Terminaisons\Poitevin"
dossier_saintongeais_faire = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Faire\Terminaisons\Saintongeais"
dossier_poitevin_avoir = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Avoir\Terminaisons\Poitevin"
dossier_saintongeais_avoir = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Avoir\Terminaisons\Saintongeais"
dossier_poitevin_etre = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Etre\Terminaisons\Poitevin"
dossier_saintongeais_etre = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Etre\Terminaisons\Saintongeais"
dossier_poitevin_dire = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Dire\Terminaisons\Poitevin"
dossier_saintongeais_dire = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Dire\Terminaisons\Saintongeais"
dossier_poitevin_voir = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Voir\Terminaisons\Poitevin"
dossier_saintongeais_voir = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Voir\Terminaisons\Saintongeais"

chemin_sortie = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Statistiques\indice_variabilité_intralocuteur.xlsx"

# Dictionnaire pour stocker les résultats combinés
resultats_combines = {}

# Dictionnaire pour stocker les dialectes des fichiers du dossier initial
dialectes_initiaux = {}

# Chemins vers les dossiers de textes pour déterminer les dialectes
dossier_poitevin_textes = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Textes locuteurs\Poitevin\Textes en txt"
dossier_saintongeais_textes = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Textes locuteurs\Saintongeais\Textes en txt"

# Parcourir les fichiers du dossier initial pour déterminer les dialectes
for nom_fichier in os.listdir(dossier_localiser):
    if nom_fichier.endswith('.txt'):
        auteur = nom_fichier.replace('.txt', '')

        if nom_fichier in os.listdir(dossier_poitevin_textes):
            dialectes_initiaux[auteur] = "Poitevin"
        elif nom_fichier in os.listdir(dossier_saintongeais_textes):
            dialectes_initiaux[auteur] = "Saintongeais"
        else:
            dialectes_initiaux[auteur] = "Inconnu"

# Fonction pour traiter un fichier et mettre à jour les résultats
def traiter_fichier(chemin_fichier):
    nom_fichier = os.path.basename(chemin_fichier)
    auteur = nom_fichier.replace('.txt', '')

    indice, total_formes, nb_variantes = calculer_indice_variabilite_et_variantes(chemin_fichier)

    if auteur not in resultats_combines:
        resultats_combines[auteur] = {
            "Auteur": auteur,
            "Indice": indice,
            "Formes": f"{total_formes}/{nb_variantes}",
            "Dialecte": dialectes_initiaux.get(auteur, "Inconnu")
        }
    else:
        # Si le fichier existe déjà, combiner les résultats
        existing_indice, existing_total_formes, existing_nb_variantes = resultats_combines[auteur]["Indice"], int(resultats_combines[auteur]["Formes"].split('/')[0]), int(resultats_combines[auteur]["Formes"].split('/')[1])

        # Combiner les indices (moyenne pondérée)
        total_combine = existing_total_formes + total_formes
        poids_existing = existing_total_formes / total_combine if total_combine != 0 else 0
        poids_nouveau = total_formes / total_combine if total_combine != 0 else 0
        indice_combine = (existing_indice * poids_existing) + (indice * poids_nouveau)

        # Mettre à jour les résultats combinés
        resultats_combines[auteur]["Indice"] = indice_combine
        resultats_combines[auteur]["Formes"] = f"{existing_total_formes + total_formes}/{existing_nb_variantes + nb_variantes}"

# Parcourir les fichiers dans chaque dossier
for dossier in [dossier_localiser, dossier_poitevin_faire, dossier_saintongeais_faire,
                dossier_poitevin_avoir, dossier_saintongeais_avoir,
                dossier_poitevin_etre, dossier_saintongeais_etre,
                dossier_poitevin_dire, dossier_saintongeais_dire,
                dossier_poitevin_voir, dossier_saintongeais_voir]:
    for nom_fichier in os.listdir(dossier):
        if nom_fichier.endswith('.txt'):
            chemin_fichier = os.path.join(dossier, nom_fichier)
            traiter_fichier(chemin_fichier)

# Convertir les résultats combinés en DataFrame
resultats = list(resultats_combines.values())
df = pd.DataFrame(resultats)

# Trier le DataFrame par la colonne "Indice" en ordre décroissant
df_sorted = df.sort_values(by="Indice", ascending=False)

# Appliquer le formatage des pourcentages
df_sorted["Indice"] = df_sorted["Indice"].apply(lambda x: f"{x:.2f}%")

# Exporter vers un fichier Excel
df_sorted.to_excel(chemin_sortie, index=False)
```

## <span style="color: #FFFFE0; font-weight: bold;">Formes attestées les plus fréquentes</span>

```python
import pandas as pd
from openpyxl import load_workbook
from openpyxl.worksheet.table import Table, TableStyleInfo
import os
import psutil
import time

# Chemins
input_path = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Flexions\FFPS2.0.tsv"
output_dir = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Statistiques"
temp_dir = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Statistiques\fichier temporaire"

os.makedirs(temp_dir, exist_ok=True)

# Tuer tous les processus Excel
for proc in psutil.process_iter():
    if proc.name() == "EXCEL.EXE":
        proc.kill()

# Lecture du TSV une seule fois
df = pd.read_csv(input_path, sep="\t", dtype=str)

# Style du tableau Excel
style = TableStyleInfo(
    name="TableStyleMedium9",
    showFirstColumn=False,
    showLastColumn=False,
    showRowStripes=True,
    showColumnStripes=False
)

def creer_fichier_frequences(colonne_lexicale, nom_fichier):
    """
    colonne_lexicale : 'form' ou 'supralemma'
    nom_fichier : nom du fichier Excel final
    """

    output_path = os.path.join(output_dir, nom_fichier)
    temp_path = os.path.join(temp_dir, f"temp_{nom_fichier}")

    # Filtrer les lignes avec fréquence
    df_filtered = df[df["frequence"].notna() & (df["frequence"].str.strip() != "")].copy()

    # Garder les colonnes nécessaires
    df_final = df_filtered[["frequence", colonne_lexicale, "source"]].copy()

    # Conversion fréquence
    df_final["frequence"] = pd.to_numeric(df_final["frequence"], errors="coerce").fillna(0).astype(int)

    # Supprimer doublons (colonne lexicale, source)
    df_final = df_final.drop_duplicates(subset=[colonne_lexicale, "source"], keep="first")

    # Écriture temporaire
    with pd.ExcelWriter(temp_path, engine="openpyxl") as writer:
        df_final.to_excel(writer, index=False, sheet_name="Données")

    # Création du tableau Excel
    wb = load_workbook(temp_path)
    ws = wb["Données"]
    table_ref = f"A1:{chr(64 + ws.max_column)}{ws.max_row}"

    table = Table(displayName="FrequencesTable", ref=table_ref)
    table.tableStyleInfo = style
    ws.add_table(table)

    wb.save(temp_path)
    wb.close()

    # Relecture pour fusion
    df_excel = pd.read_excel(temp_path)

    # Suppression source
    df_excel.drop(columns=["source"], inplace=True)

    # Fusion par colonne lexicale
    df_excel = df_excel.groupby(colonne_lexicale, as_index=False)["frequence"].sum()

    # Réécriture finale
    with pd.ExcelWriter(temp_path, engine="openpyxl") as writer:
        df_excel.to_excel(writer, index=False, sheet_name="Données")

    wb = load_workbook(temp_path)
    ws = wb["Données"]
    table_ref = f"A1:{chr(64 + ws.max_column)}{ws.max_row}"

    table = Table(displayName="FrequencesTable", ref=table_ref)
    table.tableStyleInfo = style
    ws.add_table(table)

    wb.save(temp_path)
    wb.close()

    time.sleep(2)

    if os.path.exists(output_path):
        os.remove(output_path)

    os.rename(temp_path, output_path)
    print("Fini :)")

# --------------------
# Création des deux fichiers
# --------------------

creer_fichier_frequences("form", "frequences.xlsx")
creer_fichier_frequences("supralemma", "supralemmes_frequences.xlsx")
```
