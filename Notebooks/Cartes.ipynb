## <span style="color: #FFFFE0; font-weight: bold;">Terminaisons en commun à plusieurs auteurs</span>

```python
import pandas as pd
from collections import defaultdict
from IPython.display import display

# Chemin d'accès au fichier TSV
file_path = r"C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Localiser locuteurs\occurences.tsv"

# Lire le fichier TSV
df = pd.read_csv(file_path, delimiter='\t')

# Obtenir le nombre total d'auteurs uniques
all_authors = set(auteur for auteurs in df['Auteurs'] for auteur in auteurs.split(", "))
total_authors = len(all_authors)

# Dictionnaire pour stocker les terminaisons par auteurs avec leur position
terminaison_par_auteurs = defaultdict(lambda: defaultdict(list))

# Parcourir chaque ligne du DataFrame
for _, row in df.iterrows():
    auteurs = row['Auteurs']
    terminaison = row['Terminaison']
    position = row['Position']
    temps_verbal = row['Temps verbal'].replace('*', '').strip()
    # Ajouter la terminaison avec le temps verbal à la liste des terminaisons pour cette position pour ces auteurs
    terminaison_par_auteurs[auteurs][position].append(f"{terminaison} ({temps_verbal})")

# Préparer les données pour le tableau
data = []
for auteurs, positions in terminaison_par_auteurs.items():
    terminaisons_liste = sorted(set(terminaison for terminaisons in positions.values() for terminaison in terminaisons))

    if len(terminaisons_liste) > 1:  # On s'intéresse aux groupes d'auteurs avec plus d'une terminaison
        terminaisons_liste_str = ", ".join(terminaisons_liste)
        nombre_auteurs = len(auteurs.split(", "))
        data.append({
            "Auteurs": auteurs,
            "Terminaisons": terminaisons_liste_str,
            "Nombre d'auteurs": f"{nombre_auteurs}/{total_authors}"
        })

# Créer un DataFrame pandas
results_df = pd.DataFrame(data)

# Ajuster les paramètres d'affichage de pandas
pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_rows', None)

# Afficher le tableau interactif sans l'index
display(results_df.style.hide(axis='index'))

# Enregistrer le résultat dans un fichier CSV
results_df.to_csv('terminaisons_par_position.csv', index=False)
```
