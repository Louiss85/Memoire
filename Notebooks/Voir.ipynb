## <span style="color: #FFFFE0; font-weight: bold;">Listage de mots commençant par "vu", "ve", "vo", "vay" pour chaque texte</span>

```python
import os
import re
from collections import defaultdict

# Chemins des répertoires contenant les fichiers texte
input_directories = {
    'Poitevin': r'C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Textes locuteurs\Poitevin\Textes en txt',
    'Saintongeais': r'C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Textes locuteurs\Saintongeais\Textes en txt'
}

# Chemins des répertoires où les fichiers générés seront enregistrés
output_directories = {
    'Poitevin': r'C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Voir\Pré-tri\Poitevin',
    'Saintongeais': r'C:\Users\louis\OneDrive - Université de Poitiers\Fac\Master\Mémoire\Variation flexions\Irréguliers\Voir\Pré-tri\Saintongeais'
}

# Expression régulière pour trouver les mots commençant par "vu", "ve" et "vo", "va" et ayant max 12 lettres
word_pattern = re.compile(r'\b(?:vu|ve|vé|vè|vo|va|và|v\')\w{1,10}\b', re.IGNORECASE)

# Parcourir les répertoires pour chaque dialecte
for dialect, input_directory_path in input_directories.items():
    words_by_author = defaultdict(set)

    # Parcourir tous les fichiers dans le répertoire
    for filename in os.listdir(input_directory_path):
        if filename.endswith('.txt'):
            # Extraire le nom de l'auteur du nom du fichier (ignorer les numéros)
            author = re.sub(r' \d+', '', os.path.splitext(filename)[0])
            file_path = os.path.join(input_directory_path, filename)
            with open(file_path, 'r', encoding='cp1252') as file:
                content = file.read()
                # Trouver tous les mots correspondant au motif
                words = word_pattern.findall(content)
                words = [word for word in words if not word.lower().startswith(('vein', 'vol', "ves", "veil", "ven", "vel", "von", "veul"))]
                # Ajouter les mots à l'ensemble de l'auteur
                words_by_author[author].update(words)

    # Écrire les listes de mots dans des fichiers par auteur
    output_directory_path = output_directories[dialect]
    for author, words in words_by_author.items():
        # Trier les mots par ordre alphabétique
        sorted_words = sorted(words)
        output_file_path = os.path.join(output_directory_path, f'{author}.txt')
        with open(output_file_path, 'w', encoding='utf-8') as output_file:
            for i in range(0, len(sorted_words), 10):
                line = ', '.join(sorted_words[i:i+10])
                output_file.write(line + '\n')

    print(f"Les fichiers pour le dialecte {dialect} ont été créés avec succès.")
```
